```python
import pandas as pd
import numpy as np
import matplotlib as plt

data = {
    "Current_mA": [
        -0.595,
        -0.558,
        -0.513,
        -0.479,
        -0.445,
        -0.417,
        -0.379,
        -0.345,
        -0.326,
        -0.313,
        -0.3,
        -0.289,
        -0.277,
        -0.264,
        -0.253,
        -0.238,
        -0.217,
        -0.195,
        -0.181,
        -0.162,
        -0.145,
        -0.126,
        -0.12,
        -0.1,
        -0.074,
        -0.05,
        -0.023,
        0,
        0.028,
        0.05,
        0.074,
        0.1,
        0.127,
        0.15,
        0.165,
        0.18,
        0.2,
        0.214,
        0.233,
        0.25,
        0.274,
        0.289,
        0.297,
        0.312,
        0.321,
        0.342,
        0.359,
        0.372,
        0.385,
        0.398,
        0.406,
        0.417,
        0.43,
        0.447,
        0.462,
        0.479,
        0.5,
        0.524,
        0.551,
        0.588,
    ],
    "Voltage_mV": [
        -0.0454,
        -0.0425,
        -0.0383,
        -0.0348,
        -0.0318,
        -0.0289,
        -0.0251,
        -0.0227,
        -0.0203,
        -0.0189,
        -0.018,
        -0.0174,
        -0.0168,
        -0.0156,
        -0.0135,
        -0.0115,
        -0.0094,
        -0.0088,
        -0.0079,
        -0.0078,
        -0.0061,
        -0.0041,
        -0.0016,
        0,
        0.0000,
        0,
        0.0000,
        0,
        0.0000,
        0,
        0.0000,
        0,
        0.0000,
        0.001,
        0.003,
        0.0056,
        0.008,
        0.0098,
        0.0101,
        0.01,
        0.0095,
        0.0115,
        0.0133,
        0.0151,
        0.0166,
        0.018,
        0.0183,
        0.0198,
        0.0213,
        0.0236,
        0.0251,
        0.026,
        0.0269,
        0.0284,
        0.0301,
        0.0325,
        0.0346,
        0.0372,
        0.0399,
        0.0434,
    ],
}

dfj1 = pd.DataFrame(data)
print(dfj1)

dfj1.plot(x="Current_mA", y="Voltage_mV", kind="scatter", figsize=(8, 5))

data_j2 = {
    "Current_mA": [
        -8.36,
        -7.95,
        -7.8,
        -7.18,
        -7.01,
        -6.86,
        -6.58,
        -6.43,
        -6.31,
        -6.21,
        -5.98,
        -5.78,
        -5.71,
        -5.53,
        -5.41,
        -5.09,
        -4.84,
        -4.96,
        -4.69,
        -4.39,
        -4.19,
        -4.01,
        -3.69,
        -3.44,
        -3.24,
        -3.14,
        -3.04,
        -2.79,
        -2.54,
        -2.39,
        -2.32,
        -2.14,
        -1.92,
        -1.69,
        -1.44,
        -1.35,
        -1.22,
        -1.15,
        -0.97,
        -0.7,
        -0.47,
        -0.27,
        -0.02,
        0.15,
        0.28,
        0.48,
        0.75,
        0.97,
        1.22,
        1.55,
        1.85,
        2.02,
        2.12,
        2.27,
        2.57,
        2.94,
        3.09,
        3.19,
        3.37,
        3.49,
        3.74,
        3.97,
        4.22,
        4.34,
        4.49,
        4.64,
        4.86,
        5.11,
        5.34,
        5.49,
        5.59,
        5.69,
        5.81,
        5.84,
        6.06,
        6.16,
        6.36,
        6.41,
        6.56,
        6.56,
        6.66,
        6.81,
        6.91,
        7.01,
        7.11,
        7.18,
        7.31,
        7.48,
        7.56,
        7.66,
        7.76,
        7.88,
        8.01,
        8.03,
        8.03,
        8.15,
        8.3,
        8.4,
        8.53,
        8.65,
        8.68,
        8.78,
        8.83,
        8.83,
        8.85,
        9.05,
        9.23,
        9.3,
        9.48,
        9.7,
        9.83,
        9.98,
        10.05,
        10.22,
        10.4,
        10.55,
    ],
    "Voltage_mV": [
        -0.0225,
        -0.0221,
        -0.02105,
        -0.01968,
        -0.01903,
        -0.01805,
        -0.01773,
        -0.01719,
        -0.01643,
        -0.01513,
        -0.01535,
        -0.01502,
        -0.01437,
        -0.01372,
        -0.01297,
        -0.01297,
        -0.0121,
        -0.01275,
        -0.0108,
        -0.01069,
        -0.00961,
        -0.00853,
        -0.00842,
        -0.00821,
        -0.00745,
        -0.00669,
        -0.00604,
        -0.00572,
        -0.00583,
        -0.00518,
        -0.00442,
        -0.00388,
        -0.00377,
        -0.00377,
        -0.00366,
        -0.0029,
        -0.00225,
        -0.0015,
        -0.00117,
        -0.00106,
        -0.00117,
        -0.00106,
        -0.00117,
        -0.00085,
        -0.00031,
        0.00078,
        0.00099,
        0.0011,
        0.0011,
        0.0011,
        0.0011,
        0.00207,
        0.00261,
        0.00326,
        0.00348,
        0.00359,
        0.00402,
        0.00456,
        0.00532,
        0.00575,
        0.00575,
        0.00575,
        0.00575,
        0.00619,
        0.00694,
        0.00759,
        0.00792,
        0.00813,
        0.00813,
        0.00813,
        0.00867,
        0.00943,
        0.00976,
        0.01019,
        0.01041,
        0.01041,
        0.01062,
        0.01105,
        0.0116,
        0.01116,
        0.01257,
        0.01289,
        0.01322,
        0.01365,
        0.01387,
        0.01419,
        0.01484,
        0.01506,
        0.01538,
        0.01549,
        0.01582,
        0.01614,
        0.01647,
        0.0169,
        0.01711,
        0.01733,
        0.01766,
        0.01798,
        0.01809,
        0.01874,
        0.01928,
        0.01917,
        0.01917,
        0.01917,
        0.01982,
        0.02036,
        0.02069,
        0.02166,
        0.02155,
        0.02188,
        0.02263,
        0.02339,
        0.02426,
        0.02404,
        0.02491,
        0.02501,
    ],
}

dfj2 = pd.DataFrame(data_j2)

I_min, I_max = dfj2["Current_mA"].min(), dfj2["Current_mA"].max()

dfj2["Current_mA"] = -8 + (dfj2["Current_mA"] - I_min) * (8 - (-8)) / (I_max - I_min)

print(
    dfj2["Current_mA"].min(), dfj2["Current_mA"].max()
)  # sanity check: should print -8.0 8.0
dfj2.plot(x="Current_mA", y="Voltage_mV", kind="scatter")

print(dfj2)
dfj2.plot(x="Current_mA", y="Voltage_mV", kind="scatter", figsize=(8, 5))


def extract_step_widths(df, i_col, v_col, slope_thresh, max_gradient, min_points):
    df = (
        df.groupby(i_col, as_index=False)[v_col]
        .mean()
        .sort_values(i_col)
        .reset_index(drop=True)
    )
    I = df[i_col].values
    V = df[v_col].values

    dVdI = np.gradient(V, I)  ## smoothed local slope
    point_to_point = np.diff(V) / np.diff(I)  ## raw slope between neighbouring points
    point_to_point = np.append(point_to_point, point_to_point[-1])

    flat = np.abs(dVdI) < slope_thresh

    steps = []
    start = None
    for idx in range(len(flat)):
        breaks_here = (not flat[idx]) or (np.abs(point_to_point[idx]) > max_gradient)

        if not breaks_here and start is None:
            start = idx
        if (breaks_here or idx == len(flat) - 1) and start is not None:
            end = idx if breaks_here else idx + 1
            if end - start >= min_points:
                I_seg, V_seg = I[start:end], V[start:end]
                steps.append(
                    {
                        "I_min": I_seg.min(),
                        "I_max": I_seg.max(),
                        "width": I_seg.max() - I_seg.min(),
                        "height": V_seg.mean(),
                        "n_points": len(I_seg),
                    }
                )
            start = None

    return pd.DataFrame(steps), I, V, dVdI


def plot_steps(I, V, steps_df, title):
    plt.figure(figsize=(8, 5))
    plt.plot(I, V, "o-", color="gray", ms=3, lw=1, label="data", zorder=1)

    for _, row in steps_df.iterrows():
        plt.axvspan(row["I_min"], row["I_max"], color="tab:orange", alpha=0.3, zorder=0)
        plt.axhline(row["height"], color="tab:orange", lw=0.8, alpha=0.6, zorder=0)

    plt.xlabel("Current (mA)")
    plt.ylabel("Voltage (mV)")
    plt.title(title)
    for i, row in steps_df.iterrows():
        plt.text(
            (row["I_min"] + row["I_max"]) / 2,
            row["height"],
            f"{row['width']:.3f}",
            ha="center",
            va="bottom",
            fontsize=7,
        )
    plt.legend()
    plt.show()


max_gradient1 = 2
max_gradient2 = 0.2


slope_thresh1 = 0.055
slope_thresh2 = 0.003

steps1, I1, V1, dVdI1 = extract_step_widths(
    dfj1, "Current_mA", "Voltage_mV", slope_thresh1, max_gradient1, min_points=1
)
steps2, I2, V2, dVdI2 = extract_step_widths(
    dfj2, "Current_mA", "Voltage_mV", slope_thresh2, max_gradient2, min_points=2
)

print("Junction 1")
print(steps1)
print("--" * 20)
print("Junction 2")
print(steps2)

plt.plot(I1, dVdI1, "o-")
plt.axhline(0.02, color="r", ls="--")
plt.xlabel("Current (mA)")
plt.ylabel("dV/dI")
plt.title(f"Junction 1: threshold = {slope_thresh1}")
plt.show()

plt.plot(I2, dVdI2, "o-")
plt.axhline(0.0005, color="r", ls="--")
plt.xlabel("Current (mA)")
plt.ylabel("dV/dI")
plt.title(f"Junction 2: threshold = {slope_thresh2:.2f}")

plot_steps(I1, V1, steps1, "Junction 1: I-V with detected flat steps")
plot_steps(I2, V2, steps2, "Junction 2: I-V with detected flat steps")

from scipy.special import jn


def label_steps(exp_steps_df):

    n_steps = len(exp_steps_df)
    mid = n_steps // 2

    # symmetric labels: [-k, ..., -1, 0, 1, ..., k]
    if n_steps % 2 == 1:
        labels = np.arange(-mid, mid + 1)
    else:
        labels = np.arange(-mid, mid)

    exp_steps_df = exp_steps_df.copy()
    exp_steps_df["step_label"] = labels
    return exp_steps_df


def bessel_integer_step_widths(max_n, a, scale=1.0):
    """
    determine the bessel function predicted step widths for the junction
    """
    n_vals = np.arange(-max_n, max_n + 1)
    widths = scale * np.abs(jn(np.abs(n_vals), a))
    return n_vals, widths


def compare_step_widths(exp_steps_df, a, scale=1.0):
    """
    compare the step widths of the data to the bessel function predicitons
    """
    max_n = int(exp_steps_df["step_label"].abs().max())
    n_vals, bessels = bessel_integer_step_widths(max_n, a, scale)

    exp_labels = exp_steps_df["step_label"].values
    exp_widths = exp_steps_df["width"].values

    return n_vals, bessels, exp_labels, exp_widths


def plot_bessel(exp_steps_df, a, scale=1.0, title="Symmetric Bessel comparison"):
    """
    plot the steps obtained alongside the bessel function predicted steps
    """
    n_vals, bessels, exp_labels, exp_widths = compare_step_widths(
        exp_steps_df, a, scale
    )

    plt.figure(figsize=(7, 5))
    plt.plot(exp_labels, exp_widths, "o-", lw=2, label="Experimental widths")
    plt.plot(n_vals, bessels, "s--", lw=2, label="Symmetric Bessel prediction")
    plt.xlabel("Step index n (±)")
    plt.ylabel("Step width")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()


steps1 = label_steps(steps1)
steps2 = label_steps(steps2)

# Then: compare and plot
plot_bessel(
    steps1,
    a=0.8,
    scale=steps1["width"].max(),
    title="Junction 1: Experimental vs Symmetric Bessel widths",
)

plot_bessel(
    steps2,
    a=1.2,
    scale=steps2["width"].max(),
    title="Junction 2: Experimental vs Symmetric Bessel widths",
)
```
